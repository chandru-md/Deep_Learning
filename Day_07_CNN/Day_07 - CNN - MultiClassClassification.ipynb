```python
import pandas as pd
from tensorflow import keras
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
train_data = ImageDataGenerator(rescale=1/255,shear_range=0.2,zoom_range=0.2,horizontal_flip=True,vertical_flip=True)
```

```python
test_data = ImageDataGenerator(rescale=1/255)
```

```python
train_data_generator = train_data.flow_from_directory(directory="/content/drive/MyDrive/CHANDRU M D_PROITBRIDGE/SCRIPTS/Deep_Learning/Day_07_CNN/Train_data",target_size=(256,256))
```

```python
test_data_generator = test_data.flow_from_directory(directory="/content/drive/MyDrive/CHANDRU M D_PROITBRIDGE/SCRIPTS/Deep_Learning/Day_07_CNN/Test_data",target_size=(256,256))
```

```python
from keras import Sequential
from keras.layers import Conv2D,MaxPool2D,Flatten,Dense
```

```python
model = Sequential()

model.summary()
```

```python
model = Sequential()

model.add(Conv2D(input_shape=(256,256,3),filters=8,kernel_size=(3,3),activation='relu',strides=(1,1),padding='same'))
model.add(MaxPool2D(pool_size=(2,2),strides=2))

model.add(Conv2D(filters=8,kernel_size=(3,3),activation='relu',strides=(1,1),padding='same'))
model.add(MaxPool2D(pool_size=(2,2),strides=2))

model.add(Conv2D(filters=16,kernel_size=(3,3),activation='relu',strides=(1,1),padding='same'))
model.add(MaxPool2D(pool_size=(2,2),strides=2))

model.add(Flatten())

model.add(Dense(units=30,activation='relu'))

model.add(Dense(units=4,activation='softmax')) # Changed to 4 units as there are 4 classes

model.summary()
```

```python
model.compile(optimizer='rmsprop',loss='categorical_crossentropy',metrics=['accuracy'])
```

```python
model = model.fit(x=train_data_generator,epochs=10,validation_data=test_data_generator)
```
