```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.model_selection import GridSearchCV
```

```python
def create_model():
  model = Sequential()

  model.add(Dense(units=20,input_dim=8,activation='relu',kernel_initializer='glorot_uniform'))

  model.add(Dense(units=60,activation='relu',kernel_initializer='glorot_uniform'))

  model.add(Dense(units=50,activation='relu',kernel_initializer='glorot_uniform'))

  model.add(Dense(units=1,activation='sigmoid',kernel_initializer='glorot_uniform'))

  model.summary()

  model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])

  return model
```

```python
from scikeras.wrappers import KerasClassifier
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.model_selection import GridSearchCV

def create_model(optimizer):
    model = Sequential()
    model.add(Dense(12, input_dim=x_train.shape[1], activation='relu'))
    model.add(Dense(1, activation='sigmoid'))
    model.compile(loss='binary_crossentropy', optimizer=optimizer, metrics=['accuracy'])
    return model

model = KerasClassifier(model=create_model, verbose=0)

param_grid = {
    "model__optimizer": ["adam", "rmsprop"],
    "batch_size": [16, 32],
    "epochs": [10, 20]
}

grid = GridSearchCV(estimator=model, param_grid=param_grid, cv=3, verbose=1)
grid_result = grid.fit(x_train, y_train)

print("Best Score:", grid_result.best_score_)
print("Best Params:", grid_result.best_params_)
```

```python
model=Sequential()

model.add(Dense(units=20,input_dim=8,activation='relu',kernel_initializer='glorot_uniform'))

model.add(Dense(units=60,activation='relu',kernel_initializer='glorot_uniform'))

model.add(Dense(units=50,activation='relu',kernel_initializer='glorot_uniform'))

model.add(Dense(units=1,activation='sigmoid',kernel_initializer='glorot_uniform'))

model.summary()
```

```python
neuron1=[30,50,80]
neuron2 = [40,60,90]
neuron3=[20,70,80]
activation_function = ['relu','tanh']
init=['uniform','zero']
learning_rate=[0.1,0.01,0.001]
batch_size=[10,20]
epochs=[10,12]
dropout_rate =[0.1,0.3]
```

```python
param_grid = dict(model__neuron1=[30,50,80],
model__neuron2 = [40,60,90],
model__neuron3=[20,70,80],
model__activation_function = ['relu','tanh'],
model__init=['uniform','zero'],
model__learning_rate=[0.1,0.01,0.001],
model__dropout_rate =[0.1,0.3],
batch_size=[10,20],
epochs=[10,12])
```

```python
param_grid=dict(batch_size=batch_size,epochs=epochs,learning_rate=learning_rate,dropout_rate=dropout_rate,activation_function=activation_function,init=init,neuron1=neuron1,neuron2=neuron2,neuron3=neuron3)
```

```python
def create_model(neuron1,neuron2,neuron3,activation_function,init,learning_rate,dropout_rate):
  model = Sequential()
  model.add(Dense(units=neuron1,input_dim=8,activation=activation_function,kernel_initializer='glorot_uniform'))
  model.add(Dropout(dropout_rate))
  model.add(Dense(units=neuron2,activation=activation_function,kernel_initializer=init))
  model.add(Dense(units=neuron3,activation=activation_function,kernel_initializer=init))
  model.add(Dense(units=1,activation='sigmoid',kernel_initializer='glorot_uniform'))
  model.summary()
  adam_opt=tf.keras.optimizers.Adam(learning_rate=learning_rate,beta_1=0.5,beta_2=0.96,epsilon=1e-07)
  model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
  return model
```

```python
model= KerasClassifier(model = create_model)
```

```python
grid = GridSearchCV(estimator=model,param_grid=param_grid)
```

```python
grid_result = grid.fit(x_train,y_train)
```
