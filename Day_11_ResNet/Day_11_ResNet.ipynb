```python
from tensorflow import keras
from keras.applications.vgg16 import VGG16
from keras.applications.vgg19 import VGG19
from keras.applications.resnet50 import ResNet50
from keras.applications.inception_v3 import InceptionV3
from keras.applications.convnext import ConvNeXtLarge
```

```python
VGG16_model = VGG16()

VGG16_model.summary()
```

```python
VGG19_model = VGG19()

VGG19_model.summary()
```

```python
resnet50_model = ResNet50()

resnet50_model.summary()
```

```python
convnext_model = ConvNeXtLarge()

convnext_model.summary()
```

```python
VGG_16_model = VGG16(weights='imagenet',include_top=False,input_shape=(224,224,3))

VGG_16_model.summary()
```

```python
for layer in VGG16_model.layers:
  layer.trainabel=False
```

```python
from keras.layers import Dense,Flatten
```

```python
x=Flatten()(VGG_16_model.output)
prediction=Dense(4,activation='softmax')(x)
```

```python
from tensorflow.keras.models import Model
```

```python
model=Model(inputs=VGG_16_model.inputs,outputs=prediction)
```

```python
model.summary()
```

```python
model.compile(optimizer='adam',loss='categorical_crossentropy',metrics=['accuracy'])
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```
